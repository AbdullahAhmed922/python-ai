## 1. List Operations: Removing Elements

Demonstrates removing a specific value from a list using the `.remove()` method.

```python
# List remove operation
l1 = [7, 9, 3, 1]
print("Original list:", l1)

l1.remove(3)
print("After l1.remove(3):", l1)
```

## 2. Dictionary Methods

Exploring standard dictionary methods: `.items()`, `.keys()`, `.copy()`, and `.update()`.

```python
my_dict = {'a': 1, 'b': 2, 'c': 3}

print("Items:", my_dict.items())
print("Keys:", my_dict.keys())
print("Copy:", my_dict.copy())

# Updating dictionary key 'a' to 5
my_dict.update({'a': 5})
print("Updated dictionary:", my_dict)
```

## 3. Sets: Definition and Empty Set

Notice that `{}` initializes an empty dictionary by default, while `set()` creates an empty set.

```python
s = {1, 2, 3, 4, 5}
print("Set s:", s)

e = set()
print("Empty set e:", e)
print("Type of e:", type(e))
```

## 4. Hindi-to-English Dictionary Lookup

Using `.get()` to look up keys safely without throwing a KeyError if the word is not found.

```python
words = {
    "billi": "cat",
    "kutta": "dog",
    "mota": "elephant"
}

word = input("Enter Word in Hindi: ")
# Using .get() returns a default value instead of raising KeyError
print(words.get(word, "Word not found in dictionary"))
# print(words[word])  # Note: This would raise KeyError if word is not in dictionary
```

## 5. Collecting Unique Numbers with Set

Taking numbers from the user and storing them in a set to automatically keep only unique values.

```python
s = set()

# Collect numbers from user input
for i in range(1, 8):
    n = input(f"Enter number {i}: ")
    s.add(int(n))

print("Set of unique numbers:", s)
```

## 6. Sets: Integer vs String Elements

Python sets distinguish between integer `18` and string `'18'` because they have different types and identities.

```python
s = set()
s.add(18)
s.add('18')
print("Set with int and str 18:", s)
```

## 7. Favorite Languages Dictionary

Collecting favorite programming languages for friends and storing them in a dictionary.

```python
d = {
    "Ali": "",
    "Ahmed": "",
    "Sara": "",
    "Fatima": "",
}

for friend in d:
    n = input(f"Enter your fav language {friend}: ")
    d[friend] = n

print("Dictionary of favorite languages:", d)
```

## 8. Sets and Unhashable Types

Sets cannot contain mutable items like lists (`[1, 2]`) because lists are unhashable. However, sets can be updated with items from an iterable using `.update()`.

```python
# Updating set with multiple values
s = {8, 7, 12, "Harry"}
s.update({1, 2, 3, 4, 5})
print("Updated set:", s)

# Attempting to include a mutable list in a set raises TypeError:
try:
    invalid_set = {8, 7, 12, "Harry", [1, 2]}
except TypeError as err:
    print("TypeError caught:", err)
```

## 9. Functions and Return Value

Defining a simple greeting function. Note that a bare `return` statement returns `None`.

```python
def goodday(name):
    print("Good Day " + name)
    return

a = goodday("Ali")
print("Return value of goodday('Ali'):", a)
```
