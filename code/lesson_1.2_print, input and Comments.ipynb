## 1. Output with `print()`

The `print()` function is your primary tool for debugging and communicating with the user. While it seems simple, Python provides several powerful ways to control exactly how your data is displayed.

```python
# Basic printing
print("Hello, AI world!")

# Multiple arguments (comma adds a space automatically)
print("Model:", "gemma4:31b-cloud")

# Using 'sep' to change the separator between items
print("Python", "for", "AI", sep="-")   # Output: Python-for-AI

# Using 'end' to prevent a newline (keeps the next print on the same line)
print("Thinking", end="... ")
print("done")                           # Output: Thinking... done
```

### f-Strings: Modern Python Formatting

The most efficient and readable way to embed variables into strings is using **f-strings** (formatted string literals). Simply put an `f` before the opening quote and wrap your variables in `{curly braces}`.

**Why use f-strings?**
1. They are faster than older methods.
2. They are much easier to read.
3. You can perform small calculations directly inside the braces.

```python
model = "Claude-3.5-Sonnet"
tokens = 120
cost_per_token = 0.000015

# Simple f-string
print(f"{model} model used {tokens} tokens")

# f-string with calculations
print(f"Total estimated cost: ${tokens * cost_per_token:.4f}")
```

### 💡 AI Context: Multi-line Strings
In AI development, we often work with long system prompts. We use **triple quotes** (`"""`) to create multi-line strings that preserve formatting.

## 2. Input with `input()`

To make a program interactive, we use the `input()` function. This allows the user to type text into the program, which Python then stores as a variable.

```python
name = input("Enter your name: ")
print(f"Welcome to the AI Course, {name}! 🚀")
```

### ⚠️ The Golden Rule of `input()`

**The `input()` function ALWAYS returns a string (`str`), even if the user types a number.**

If you want to use the input for math, you must use **Type Casting** to convert the string into an integer (`int`) or a decimal (`float`).

```python
# Scenario: User enters age
age_str = input("Enter your age: ")
print(f"Type of age_str: {type(age_str)}") # Always <class 'str'>

# Convert to integer for calculations
age_int = int(age_str)
print(f"Type of age_int: {type(age_int)}") # Now <class 'int'>

# Now we can do math!
years_until_100 = 100 - age_int
print(f"You will be 100 in {years_until_100} years!")
```

## 4. Mini Build: The Prompt Engineering Tool 🛠️

Let's combine everything we've learned: `input()`, `f-strings`, and `comments` to build a tool that helps you generate high-quality prompts for AI models.

```python
# --- Input Section ---
print("Welcome to the AI Prompt Generator!")
role = input("Define the AI Role (e.g., Senior Python Developer): ")
topic = input("What is the topic to explain? ")
level = input("Target Audience Level (e.g., Beginner, Executive): ")
points = input("Number of bullet points required: ")

# Type Casting: Ensure points is an integer
try:
    points_int = int(points)
except ValueError:
    print("Invalid number entered. Defaulting to 3 points.")
    points_int = 3

# --- Logic Section ---
# Use an f-string to construct the final system prompt
final_prompt = f"You are a {role}. Please explain {topic} to a {level} using exactly {points_int} clear bullet points."

# --- Output Section ---
print("\n" + "="*30)
print("🚀 YOUR GENERATED PROMPT 🚀")
print("="*30)
print(final_prompt)
print("="*30)
```
